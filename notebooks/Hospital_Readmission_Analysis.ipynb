## 5. Readmission Distribution 

```python
import matplotlib.pyplot as plt 


readmission_counts = df["readmitted"].value_counts()

plt.figure(figsize = (8,5))


bars = plt.bar(
    readmission_counts.index,
    readmission_counts.values
)


for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height() + 500,
        f"{int(bar.get_height()):,}",
        ha="center",
        fontsize=10
    )

plt.title("Distribution of Hospital Readmission Status", fontsize=16)
plt.xlabel("Readmission Category", fontsize=12)
plt.ylabel("Number of Hospital Encounters", fontsize=12)

plt.tight_layout()
plt.show()
```

## 6. Readmission by age group

```python
pd.crosstab(df["age"], df["readmitted"])
```

```python
readmission_age = pd.crosstab(df["age"], df["readmitted"])

readmission_age.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Hospital Readmissions by Age Group", fontsize=16)
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## 7. Laboratory prodecures by readmission

```python
lab = (
    df.groupby("readmitted")["num_lab_procedures"]
      .mean()
      .sort_values(ascending=False)
)

lab
```

```python
import matplotlib.pyplot as plt

lab.plot(kind="bar")

plt.title("Average Number of Laboratory Procedures by Readmission Status")
plt.xlabel("Readmission")
plt.ylabel("Average Number of Lab Procedures")
plt.xticks(rotation=0)

plt.show()
```

### Insight

Patients readmitted within 30 days underwent the highest average number of laboratory procedures (44.23), followed closely by patients readmitted after 30 days (43.84). Patients who were not readmitted had the lowest average (42.38).

This suggests that patients requiring more diagnostic testing during hospitalization may have more complex medical conditions and could be at greater risk of future readmission.

## 10. Readmission rate by age group 

```python
age_readmission = (
    pd.crosstab(df["age"], df["readmitted"], normalize="index") * 100
)

age_readmission
```

```python
import matplotlib.pyplot as plt

age_readmission[["<30", ">30", "NO"]].plot(kind="bar", figsize=(12,6))

plt.title("Readmission Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Percentage of Patients")
plt.xticks(rotation=45)

plt.legend(title="Readmission")
plt.show()
```

### Insight

Readmission rates vary considerably across age groups. Patients aged 70–90 years experienced some of the highest readmission rates, particularly for readmissions occurring after 30 days. In contrast, children aged 0–10 years had the lowest readmission rates and the highest proportion of patients who were not readmitted.

These findings suggest that older patients may require additional discharge planning, follow-up care, and chronic disease management to reduce the likelihood of returning to the hospital.

## 13. Medical speciality and readmission

```python
specialty = (
    pd.crosstab(
        df["medical_specialty"],
        df["readmitted"],
        normalize="index"
    ) * 100
)

specialty
```

```python
top_specialties = (
    df[df["medical_specialty"] != "?"]["medical_specialty"]
      .value_counts()
      .head(20)
      .index
)

specialty_top = specialty.loc[top_specialties]

specialty_top
```

```python
import matplotlib.pyplot as plt

specialty_top["<30"].sort_values().plot(
    kind="barh",
    figsize=(9,6)
)

plt.title("30-Day Readmission Rate by Medical Specialty")
plt.xlabel("Percentage of Patients Readmitted Within 30 Days")
plt.ylabel("Medical Specialty")

plt.show()
```

### Insight

Thirty-day readmission rates varied considerably across medical specialties. Oncology patients had the highest proportion of 30-day readmissions among the most common specialties, followed by Nephrology and Physical Medicine & Rehabilitation. In contrast, Obstetrics & Gynecology and several surgical specialties had the lowest readmission rates.

These findings suggest that patients treated in specialties managing chronic or medically complex conditions are more likely to experience early readmission. Hospitals could prioritize transitional care, discharge planning, and follow-up programs within these specialties to reduce avoidable readmissions.

## 14. Readmission by race

```python
race_readmission = (
    pd.crosstab(
        df["race"],
        df["readmitted"],
        normalize="index"
    ) * 100
)

race_readmission = race_readmission.drop("?", errors="ignore")

race_readmission
```

```python
import matplotlib.pyplot as plt

race_readmission["<30"].sort_values().plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("30-Day Readmission Rate by Race")
plt.xlabel("Race")
plt.ylabel("Percentage of Patients Readmitted Within 30 Days")
plt.xticks(rotation=45)

plt.show()
```

### Insight

Thirty-day readmission rates were relatively similar across racial groups, ranging from approximately 9.6% to 11.3%. Caucasian and African American patients showed slightly higher readmission rates than the other groups included in the dataset, although the differences were modest.

These findings suggest that race alone does not appear to be a strong predictor of readmission in this dataset. Additional factors such as patient health status, chronic conditions, hospital utilization, and treatment complexity are likely to have a greater influence on readmission risk.

## 15. Readmission by gender 

```python
gender_readmission = (
    pd.crosstab(
        df["gender"],
        df["readmitted"],
        normalize="index"
    ) * 100
)

gender_readmission = gender_readmission.drop("Unknown/Invalid", errors="ignore")

gender_readmission
```

```python
import matplotlib.pyplot as plt

gender_readmission["<30"].sort_values().plot(
    kind="bar",
    figsize=(6,4)
)

plt.title("30-Day Readmission Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Percentage of Patients Readmitted Within 30 Days")
plt.xticks(rotation=0)

plt.show()
```

### Insight

Female and male patients had nearly identical 30-day readmission rates, with females showing a slightly higher rate (11.25%) than males (11.06%). The small difference suggests that gender alone is not a strong predictor of hospital readmission in this dataset.

Other clinical factors, such as disease severity, previous hospital utilization, and treatment complexity, are likely to play a much greater role in predicting readmission risk.

## 16. Average length of stay by age group 

```python
los_age = (
    df.groupby("age")["time_in_hospital"]
      .mean()
)

los_age
```

```python
import matplotlib.pyplot as plt

los_age.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Average Length of Hospital Stay by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average Days in Hospital")
plt.xticks(rotation=45)

plt.show()
```

### Insight

The average length of hospital stay increased steadily with patient age. Younger patients (0–10 years) stayed approximately 2.5 days on average, whereas patients aged 80–90 years stayed nearly 4.8 days.

This trend suggests that older patients often require longer hospitalization due to more complex medical conditions, multiple chronic diseases, and longer recovery periods. Longer hospital stays among older adults also emphasize the importance of discharge planning and post-hospital follow-up to reduce the likelihood of future readmissions.

## 17. Correlation between numerical variables

```python
numeric_df = df.select_dtypes(include="number")

correlation = numeric_df.corr()

correlation
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,10))

plt.imshow(correlation, cmap="coolwarm", aspect="auto")

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=90,
    fontsize=8
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns,
    fontsize=8
)

plt.title("Correlation Heatmap of Numerical Variables")

plt.tight_layout()
plt.show()
```

### Insight

The correlation heatmap shows that most numerical variables have weak to moderate relationships, indicating that no single factor alone strongly predicts patient outcomes.

The strongest positive correlation was observed between the length of hospital stay (time_in_hospital) and the number of medications (num_medications), suggesting that patients who remain hospitalized longer generally receive more medications during their stay.

A moderate positive relationship was also found between the length of stay and the number of laboratory procedures (num_lab_procedures), indicating that patients requiring more diagnostic testing tend to spend more time in the hospital.

Overall, the findings suggest that hospital utilization variables, including length of stay, medications, laboratory procedures, and prior hospital visits, are more closely related to one another than demographic characteristics, highlighting the complexity of predicting hospital readmission.

# Conclusion

This project explored factors associated with hospital readmission using the Diabetes 130-US Hospitals dataset containing over 100,000 patient encounters.

The analysis found that patients who were readmitted generally experienced greater healthcare utilization than patients who were not readmitted. Readmitted patients had longer hospital stays, underwent more laboratory procedures, received more medications, and had more previous inpatient and emergency visits.

Age also played an important role. Older patients tended to stay in the hospital longer, while younger patients generally had shorter admissions. Certain medical specialties, such as Oncology and Nephrology, showed higher 30-day readmission rates, suggesting that patients treated for more complex conditions may require additional post-discharge care.

In contrast, demographic characteristics such as gender and race showed relatively small differences in readmission rates, indicating that clinical and healthcare utilization factors appear to be stronger indicators of readmission than demographic variables alone.

Overall, this exploratory analysis demonstrates how healthcare data analytics can be used to identify patterns associated with hospital readmissions and provide insights that may support better resource allocation, discharge planning, and patient care strategies.
